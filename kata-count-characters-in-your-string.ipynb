## El encargo

La radio de la base recibe mensajes y hay que hacer el **inventario de caracteres**: un diccionario que diga cuántas veces aparece cada letra.

Recibes un string y devuelves un diccionario `{letra: cuenta}`. Si el string está vacío, devuelves un diccionario vacío.

Ejemplos:
- `"aba"` → **`{"a": 2, "b": 1}`**
- `""` → **`{}`**
- `"aab"` → **`{"a": 2, "b": 1}`**

Firma: `def count(texto)`.

## Hoja de ruta

1. **Entrada:** un string cualquiera.
2. **Salida:** un diccionario `{letra: cuántas veces}`.
3. **Es `contar` del capítulo, tal cual:** diccionario vacío, `for` por los caracteres, y por cada uno sumar 1 a su cuenta.
4. **La clave del `if`:** si la letra aún no está, `cuentas[letra] = 1`; si ya está, `cuentas[letra] += 1`. O `cuentas[letra] = cuentas.get(letra, 0) + 1` en una línea.
5. **Casos límite:** `""` → `{}` (el bucle no da ni una vuelta); cada carácter cuenta, incluidos espacios y signos.

### Versión de referencia (comentada)

```python
def count(texto):
    cuentas = {}
    for letra in texto:
        cuentas[letra] = cuentas.get(letra, 0) + 1
    return cuentas

# con el if desplegado, como sale en el capítulo:
def count_if(texto):
    cuentas = {}
    for letra in texto:
        if letra not in cuentas:
            cuentas[letra] = 1
        else:
            cuentas[letra] += 1
    return cuentas
```

**La moraleja:** Este kata ES el ejemplo `contar` del capítulo. La única decisión real es la línea del `get`: «dame la cuenta que lleva esta letra, y si no está, 0». Todo lo demás es inicializar, recorrer y devolver.
