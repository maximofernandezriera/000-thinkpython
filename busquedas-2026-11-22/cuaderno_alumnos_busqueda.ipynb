## 2. Búsqueda binaria

Si la lista está **ordenada**, no hace falta mirarlo todo: apunta al centro y descarta la mitad que no puede contener el objetivo.

Tres variables controlan la «zona viva» de la lista:

- `low`: primer índice de la zona (empieza en `0`)
- `high`: último índice de la zona (empieza en `len(lista) - 1`)
- `mid`: el centro de la zona, `(low + high) // 2`

Mientras `low <= high`: mira `lista[mid]`. Si es el objetivo, devuelve `mid`. Si te has pasado, baja `high` a `mid - 1`. Si te has quedado corto, sube `low` a `mid + 1`. Si la zona se vacía, devuelve `None`.

Traza buscando `40` en `[2, 3, 4, 13, 40]`:

| paso | low | high | mid | lista[mid] | ¿qué pasa? |
|---|---|---|---|---|---|
| 1 | 0 | 4 | 2 | 4 | corto → `low = 3` |
| 2 | 3 | 4 | 3 | 13 | corto → `low = 4` |
| 3 | 4 | 4 | 4 | 40 | ¡encontrado! devuelve 4 |

## Resumen

- **Secuencial**: simple, sirve aunque la lista no esté ordenada, hasta n miradas.
- **Binaria**: exige orden; hasta ⌊log₂ n⌋ + 1 miradas para n ≥ 1 (0 si la lista está vacía); la zona viva se estrecha con `low` y `high`.
- Contar comparaciones enseña la diferencia mejor que medir tiempos.
