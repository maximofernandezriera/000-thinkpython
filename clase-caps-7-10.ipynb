## Parte 7 — Iteración y búsqueda (capítulo 7)

### El bucle `while`: repetir mientras se cumple una condición

La cuenta atrás del despegue no sabemos hacerla con un `for` cómodo todavía: mejor un `while`, que repite **mientras** la condición sea cierta.

Predecid: ¿qué imprime esto?

```python
def cuenta_atras(n):
    while n > 0:
        print(n)
        n = n - 1
    print("Despegue.")

cuenta_atras(3)
```

Tres piezas que hacen que el `while` termine:

1. **Inicializar** antes del bucle: `n` ya vale algo.
2. **Condición** en la cabecera: `n > 0`.
3. **Actualizar** dentro: `n = n - 1` acerca `n` a la salida.

Si falta la tercera, el bucle no termina nunca: la condición se cumple siempre.

### `break`: salir antes de tiempo

A veces queremos interrumpir aunque la condición siga siendo cierta: `break` rompe el bucle en el acto.

```python
def revisar_oxigeno(lecturas):
    for lectura in lecturas:
        if lectura < 20:
            print("Alerta con", lectura)
            break
        print("ok:", lectura)

revisar_oxigeno([95, 80, 60, 15, 90])
```

### `while` frente a `for`

- `for x in coleccion`: recorro elemento a elemento; sé cuántas vueltas como mucho.
- `while condicion`: repito mientras algo se cumpla; sirve cuando **no sé cuántas vueltas** harán falta.

Todo `for` se puede escribir como `while`... y casi nunca merece la pena.

### Newton: la raíz cuadrada sin `sqrt`

El libro calcula raíces cuadradas mejorando una estimación hasta que deja de cambiar. No sabemos cuántas vueltas hace falta: por eso es un `while`.

```python
def raiz(a):
    x = a / 2
    while True:
        y = (x + a / x) / 2
        if y == x:
            return x
        x = y

print(raiz(4))
print(raiz(9))
print(raiz(2))
```

### Búsqueda: recorrer hasta encontrar (o no)

La estación envía lista de muestras recogidas; queremos saber si ya tenemos una. Con una lista, la única forma es **recorrerla**.

```python
def tiene_muestra(muestras, objetivo):
    for muestra in muestras:
        if muestra == objetivo:
            return True
    return False

muestras = ["basalto", "olivino", "hematita"]
print(tiene_muestra(muestras, "olivino"))
print(tiene_muestra(muestras, "hielo"))
```

Fijaos en la posición de los `return`:

- `return True` **dentro** del bucle: en cuanto aparece, salimos.
- `return False` **fuera**: solo si el bucle terminó sin encontrarla.

Si el `return False` se indenta dentro del bucle, la función solo mira el primer elemento — error típico.

### Búsqueda que devuelve la posición

Variante: no «sí o no», sino **dónde** está. Si no está, devolvemos `-1`.

```python
def posicion_muestra(muestras, objetivo):
    for i in range(len(muestras)):
        if muestras[i] == objetivo:
            return i
    return -1

muestras = ["basalto", "olivino", "hematita"]
print(posicion_muestra(muestras, "hematita"))
print(posicion_muestra(muestras, "hielo"))
```

### Una colección de contadores

El patrón estrella del capítulo: contar cuántas veces aparece cada cosa. El rover clasifica muestras y queremos el recuento por tipo.

```python
def contar(secuencia):
    contador = {}
    for x in secuencia:
        if x not in contador:
            contador[x] = 1
        else:
            contador[x] += 1
    return contador

muestras = ["basalto", "olivino", "basalto", "hematita", "olivino", "basalto"]
print(contar(muestras))
print(contar("banana"))
```

Y la versión con `get`, sin `if`: `contador[x] = contador.get(x, 0) + 1` — «la cuenta que lleva, y si no hay, 0».

### Recorrer un diccionario

Un `for` sobre un diccionario recorre **las claves**. El valor se pide aparte con `d[clave]`.

```python
inventario = {"oxigeno": 120, "agua": 340, "comida": 85}
for recurso in inventario:
    print(recurso, "->", inventario[recurso])
```

### Filtrar con diccionario: recursos por debajo del mínimo

Juntamos los dos capítulos: recorrer las claves (cap. 10) con el patrón recorrer-y-recoger (cap. 7).

```python
def bajo_minimos(inventario, limite):
    alerta = []
    for recurso in inventario:
        if inventario[recurso] < limite:
            alerta.append(recurso)
    return alerta

inventario = {"oxigeno": 120, "agua": 300, "comida": 85}
print(bajo_minimos(inventario, 150))
```

### Memos: no repetir cálculos

La versión rápida de Fibonacci del capítulo guarda en un diccionario los resultados ya calculados.

```python
conocidos = {0: 0, 1: 1}

def fibonacci_memo(n):
    if n in conocidos:
        return conocidos[n]
    resultado = fibonacci_memo(n - 1) + fibonacci_memo(n - 2)
    conocidos[n] = resultado
    return resultado

print(fibonacci_memo(30))
print(len(conocidos), "valores guardados")
```

### Depuración: claves que se escapan

- `d[clave]` explota si la clave no está: comprobad con `in` o usad `get`.
- `for x in d` recorre **claves**, no valores ni parejas.
- Un diccionario no tiene orden de posiciones: `d[0]` no es «el primero», es la clave `0`.
- Las claves deben ser inmutables: strings, enteros, tuplas. **Listas no.**
