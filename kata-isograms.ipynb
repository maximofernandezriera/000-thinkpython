## El encargo

Un **isograma** es una palabra (o frase) sin letras repetidas. Para el control de carga de la base sirve para detectar códigos duplicados.

Recibes un string y devuelves `True` si es un isograma, `False` si alguna letra aparece más de una vez. **No distingue mayúsculas** y el string vacío también cuenta como isograma.

Ejemplos:
- `"dermatoglyphics"` → **True**
- `"aba"` → **False** (la `a` se repite)
- `"moOse"` → **False** (la `o` se repite, aunque una sea mayúscula)
- `""` → **True**

Firma: `def is_isogram(palabra)`.

## Hoja de ruta

1. **Entrada:** un string cualquiera (puede estar vacío).
2. **Salida:** un booleano.
3. **Mayúsculas primero:** `palabra.lower()` antes de nada, como en los ejemplos del capítulo.
4. **Contar con diccionario:** el patrón `contar` del capítulo — un diccionario vacío, un `for` por las letras y `cuentas[letra] = cuentas.get(letra, 0) + 1`.
5. **Después, recorrer las claves:** si alguna cuenta es mayor que 1, `False`.
6. **Atajo:** si una letra ya está en el diccionario cuando llegas a ella, puedes devolver `False` en el acto.
7. **Casos límite:** `""` → `True`; una sola letra → `True`.

### Versión de referencia (comentada)

```python
def is_isogram(palabra):
    cuentas = {}
    for letra in palabra.lower():   # minúsculas primero
        if letra in cuentas:        # ¿ya la vimos?
            return False
        cuentas[letra] = 1          # primera vez que aparece
    return True

# con set: el mismo criterio en una línea (los sets no repiten)
def is_isogram_set(palabra):
    palabra = palabra.lower()
    return len(palabra) == len(set(palabra))
```

**La moraleja:** `letra in cuentas` usa el diccionario como «memoria de lo visto»: la primera vez anotas, la segunda detectas la repetición. Es la mitad del patrón contar, con salida anticipada.
