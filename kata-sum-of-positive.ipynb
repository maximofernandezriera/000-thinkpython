### Versión de referencia (comentada)

```python
def positive_sum(lista):
    total = 0
    for n in lista:          # recorrer la lista
        if n > 0:            # el filtro
            total += n       # el acumulador
    return total

# una línea, mismo recorrido:
def positive_sum_corta(lista):
    return sum(n for n in lista if n > 0)
```

**La moraleja:** El patrón recorrer-filtrar-acumular sale en la mitad de los katas de listas. Fíjate en las tres piezas: `total = 0` antes del bucle, `if` dentro, `return` fuera.
