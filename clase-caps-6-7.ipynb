### return no es print

Ya lo sospechábamos desde Codewars, pero hoy lo vemos en serio. `print` enseña algo en pantalla; `return` entrega un valor que se puede guardar, comparar, seguir usando. Una función sin `return` devuelve `None`.

```python
def velocidad_mala(v):
    print(v * 2)

def velocidad_buena(v):
    return v * 2

x = velocidad_mala(40000)
x   # imprime 80000... pero x no guarda nada
```

```python
x = velocidad_buena(40000)
x + 1000
```

```python
type(velocidad_mala(10))   # None: el valor "que no es un valor"
```

Y ojo a este clásico: todo lo que va **después** de un `return` dentro de la misma rama no se ejecuta nunca. Código muerto.

```python
def inutil():
    return 42
    print("esto no llega a verse")   # código muerto

inutil()
```

### Desarrollo incremental

Consejo del capítulo que vale oro: no escribáis la función entera de golpe. Probadla por trozos. Lo hacemos en vivo calculando la distancia real entre la Tierra y Marte dadas sus coordenadas.

```python
# Paso 1: solo la declaración y una comprobación de que entran los datos
def distancia(x1, y1, x2, y2):
    return 0   # todavía no, pero ya podemos llamarla

distancia(0, 0, 3, 4)
```

```python
# Paso 2: calculamos las diferencias y las mostramos antes de seguir
def distancia(x1, y1, x2, y2):
    dx = x2 - x1
    dy = y2 - y1
    print("dx:", dx, "dy:", dy)   # print temporal, de depuración
    return 0

distancia(0, 0, 3, 4)
```

```python
# Paso 3: ya confiamos en dx y dy, aplicamos Pitágoras
import math

def distancia(x1, y1, x2, y2):
    dx = x2 - x1
    dy = y2 - y1
    return math.sqrt(dx**2 + dy**2)

distancia(0, 0, 3, 4)   # triángulo 3-4-5, sabemos que debe dar 5
```

Eso es el **desarrollo incremental**: poco código, probar, un poco más, probar otra vez. Y al probar elegimos valores cuyo resultado ya conocemos (el 3-4-5 de toda la vida).

### Funciones booleanas

Una función que devuelve `True` o `False` se llama **booleana**, y se usan como condiciones de `if`. Aquí hay una evolución típica en tres pasos:

```python
# versión 1: explícita
def es_par(n):
    if n % 2 == 0:
        return True
    else:
        return False

es_par(10)
```

```python
# versión 2: la comparación YA vale True o False, no hace falta el if
def es_par(n):
    return n % 2 == 0

es_par(10), es_par(7)
```

```python
# y una booleana se usa directamente como condición
def nivel_alerta(litros):
    if es_par(litros):          # se lee como inglés: "si es par"
        return "nivel par"
    return "nivel impar"

nivel_alerta(42)
```

### Recursión que devuelve

La recursión del capítulo 5 imprimía; ahora devuelve. El factorial otra vez, pero fijaos en la confianza: asumimos que `factorial(n-1)` funciona y solo resolvemos el caso de `n`. El libro lo llama el **salto de fe**.

```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

factorial(6)
```

```python
# fibonacci: cada término es la suma de los dos anteriores
def fibonacci(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2)

fibonacci(10)
```

### Comprobar los tipos de entrada

¿Qué pasa si alguien llama a `factorial` con algo que no es un entero?

```python
# esto va a fallar: factorial de un decimal nunca llega al caso base
factorial(1.5)   # RecursionError
```

`1.5 - 1 = 0.5`, `-0.5`... nunca toca el `0` exacto. La cura es validar la entrada con `isinstance` al principio:

```python
def factorial(n):
    if not isinstance(n, int):
        print("factorial solo admite enteros")
        return None
    if n < 0:
        print("factorial no admite negativos")
        return None
    if n == 0:
        return 1
    return n * factorial(n - 1)

factorial(1.5)
factorial(-3)
factorial(6)
```

Estos guardianes al principio de la función se llaman *precondiciones*: si la entrada no vale, se avisa pronto y claro, en lugar de explotar 40 líneas más abajo en un traceback críptico.

## Parte 7 — Iteración y búsqueda (capítulo 7)

### Recorrer un string

Un `for` puede recorrer un string letra a letra. Houston nos manda telemetría como texto; vamos a inspeccionarla carácter a carácter.

```python
mensaje = "ORBITA-ESTABLE"

for letra in mensaje:
    print(letra, end=' ')
```

`end=' '` hace que `print` no salte de línea. Ahora la misma idea empaquetada en una función que **responde**, no que solo imprime:

```python
def contiene_e(palabra):
    for letra in palabra:
        if letra == 'e' or letra == 'E':
            return True     # en cuanto la encuentra, se acabó
    return False            # solo llega aquí si recorrió todo sin éxito

contiene_e("ORBITA"), contiene_e("MARTE")
```

Fijaos en la estructura, que se repite mil veces en programación: dentro del bucle un `return True` que corta en seco, y el `return False` **fuera** del bucle, al final. Si lo metes dentro del bucle solo miras la primera letra — probadlo mentalmente.

### Actualizar variables: el contador y el acumulador

Para contar cosas hace falta una variable que cambia en cada vuelta. Inicializar antes del bucle, actualizar dentro.

```python
mensaje = "ORBITA-ESTABLE"
contador = 0

for letra in mensaje:
    if letra == 'A':
        contador = contador + 1

contador
```

```python
# la forma abreviada, que es la que usa todo el mundo
contador = 0
for letra in mensaje:
    if letra == 'A':
        contador += 1
contador
```

`contador += 1` es exactamente `contador = contador + 1`. Este patrón — *inicializar, recorrer, actualizar* — es la mitad de los bucles que escribiréis en vuestra vida.

### La lista de palabras del libro

El capítulo 7 trabaja con una lista real de 113.809 palabras en inglés. La descargamos igual que hace el libro:

```python
from os.path import basename, exists

def descargar(url):
    archivo = basename(url)
    if not exists(archivo):
        from urllib.request import urlretrieve
        urlretrieve(url, archivo)
    return archivo

descargar('https://raw.githubusercontent.com/AllenDowney/ThinkPython/v3/words.txt')
```

```python
# open() abre el archivo; iterar sobre él da línea a línea
# strip() quita el salto de línea del final
total = 0
for linea in open('words.txt'):
    palabra = linea.strip()
    total += 1

total
```

```python
# ¿qué porcentaje de palabras inglesas no tienen 'e'? (el reto de Gadsby)
sin_e = 0
for linea in open('words.txt'):
    palabra = linea.strip()
    if not contiene_e(palabra):
        sin_e += 1

print(f"{sin_e} de {total}: un {sin_e / total * 100:.1f}% sin 'e'")
```

### Búsqueda más fina: uses_any

Generalizamos: ¿usa la palabra alguna de unas letras dadas? Misma estructura, dos strings en lugar de letra y letra:

```python
def usa_alguna(palabra, letras):
    for letra in palabra.lower():
        if letra in letras.lower():
            return True
    return False

usa_alguna("banana", "aeiou")
```

```python
usa_alguna("ritmo", "aeiou")   # ¿palabra sin vocales?
```

```python
# y al revés: palabras que evitan unas letras (los "lipogramas")
def evita(palabra, prohibidas):
    for letra in palabra.lower():
        if letra in prohibidas.lower():
            return False
    return True

evita("ritmo", "aeiou"), evita("luna", "aeiou")
```

```python
# búsqueda en la lista: ¿cuántas palabras inglesas evitan las vocales?
raras = 0
for linea in open('words.txt'):
    palabra = linea.strip()
    if evita(palabra, "aeiou"):
        raras += 1

raras
```

### Doctest: pruebas que viven en el docstring

Último regalo del capítulo: los ejemplos escritos en el docstring con `>>>` se pueden **ejecutar como pruebas automáticas**. Documentación y tests en el mismo sitio.

```python
def usa_alguna(palabra, letras):
    """True si la palabra usa alguna de las letras dadas.

    >>> usa_alguna('banana', 'aeiou')
    True
    >>> usa_alguna('ritmo', 'aeiou')
    False
    """
    for letra in palabra.lower():
        if letra in letras.lower():
            return True
    return False

import doctest
doctest.testmod(verbose=True)
```

Si algún día rompemos la función al tocarla, `testmod` nos avisa. Tests gratis dentro de la documentación.
