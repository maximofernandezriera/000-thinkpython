### Versión de referencia (comentada)

```python
def disemvowel(texto):
    resultado = ""                         # string nuevo: los strings son inmutables
    for letra in texto:
        if letra not in "aeiouAEIOU":      # filtro con not in
            resultado = resultado + letra  # concatenar crea otro string
    return resultado

# con expresiones regulares (capítulo 8):
import re

def disemvowel_re(texto):
    return re.sub("[aeiouAEIOU]", "", texto)   # cada vocal se sustituye por ""
```

**La moraleja:** Hay dos formas de verlo. Con el bucle construyes un string nuevo letra a letra, porque el original no se puede tocar. Con `re.sub` describes el patrón (`[aeiouAEIOU]` es "cualquiera de estas letras") y lo sustituyes por el string vacío.
