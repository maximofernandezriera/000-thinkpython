## 1 · La idea en una frase

- `print` **muestra** algo en pantalla (efecto para el ojo humano).
- `return` **entrega** un valor al código que llamó a la función (dato para el programa).

Una función sin `return` devuelve `None` implícitamente.

### 1.1 — Devolver vs imprimir

Aquí la parte clave: `b` queda en `None` aunque la función imprimiera 25.

```python
def cuadrado(x):
    return x * x          # entrega el valor

def cuadrado_show(x):
    print(x * x)          # solo lo muestra

a = cuadrado(5)
b = cuadrado_show(5)
print("a =", a)
print("b =", b)           # b es None: no devolvió nada
```

## 2 · `return` termina la función inmediatamente

Todo lo que haya **debajo** de un `return` ejecutado no corre nunca. Sirve para salir pronto de una función ("retorno temprano").

### 2.1 — El código muerto

En PythonTutor verás que para n=-3 nunca se evalúa el segundo `return`.

```python
def prueba(n):
    if n < 0:
        return "negativo"   # sale aquí; lo de abajo no corre
    return "positivo"

print(prueba(-3))
print(prueba(10))
```

## 3 · Varios `return` en una función

No hay un solo `return` válido: cada rama puede devolver algo distinto. Python ejecuta el primero que alcance.

### 3.1 — if/elif/else con tres salidas

Mira en PythonTutor cómo en cada llamada solo se ejecuta UN `return`.

```python
def clasifica(n):
    if n < 0:
        return "negativo"
    elif n == 0:
        return "cero"
    else:
        return "positivo"

for v in [-7, 0, 42]:
    print(v, "->", clasifica(v))
```

### 3.2 — Devolver booleanos directamente

Patrón típico de los katas: no hace falta `if ... return True else return False`.

```python
def es_par(n):
    return n % 2 == 0      # la comparación YA es True/False

print(es_par(4))
print(es_par(7))
```

## 4 · `return` sin valor y `None`

`return` a secas sale de la función devolviendo `None`. Útil para cortar la ejecución temprano cuando la función es de "efecto" (imprime, dibuja...).

### 4.1 — Salida temprana con return a secas

```python
def cuenta_atras(n):
    if n <= 0:
        print("¡nada que contar!")
        return              # devuelve None y termina
    for i in range(n, 0, -1):
        print(i)
    print("¡ya!")

cuenta_atras(0)
cuenta_atras(3)
```

El clásico error de los alumnos: guardan el 'resultado' de una función que solo imprime.

```python
def saluda():
    print("hola")

resultado = saluda()
print("resultado =", resultado)   # None
```

## 5 · Devolver varias cosas: tuplas

Un `return` puede entregar varios valores empaquetados en una tupla.

### 5.1 — Dos valores en un return

En PythonTutor verás la tupla viajar de la función a las dos variables.

```python
def division_resto(a, b):
    cociente = a // b
    resto = a % b
    return cociente, resto    # una tupla (cociente, resto)

c, r = division_resto(17, 5)
print("cociente:", c)
print("resto:", r)
```

## 6 · El flujo completo, en un ejemplo mayor

Funciones que llaman a funciones: cada `return` entrega a su llamante, y los marcos apilan y desaparecen. Este es EL ejemplo para ver despacio en PythonTutor.

### 6.1 — Funciones componiendo con return

Pulsa Next en PythonTutor y observa: aparece un marco nuevo para `area_cuadrado`, devuelve 16, su marco desaparece, y `area_dos_cuadrados` continúa con ese valor.

```python
def area_cuadrado(lado):
    return lado * lado

def area_dos_cuadrados(lado):
    area = area_cuadrado(lado)     # llama a la otra función
    return area * 2

total = area_dos_cuadrados(4)
print("total =", total)
```
