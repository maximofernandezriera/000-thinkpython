## El encargo

Recibes un texto y debes devolver **cuántas vocales** tiene. Para este kata, las vocales son `a`, `e`, `i`, `o` y `u` (la `y` no cuenta).

El texto solo tiene letras minúsculas y espacios.

Ejemplos:
- `"abracadabra"` → **5**
- `"rover en marte"` → **5**
- `"xyz"` → **0**

Firma: `def get_count(texto)`.

## Hoja de ruta

1. **Entrada:** un string en minúsculas (puede tener espacios o estar vacío).
2. **Salida:** un entero, el número de vocales.
3. **Recorrido:** `for letra in texto:` visita cada letra, igual que `has_e` en el capítulo.
4. **¿Cómo sé si una letra es vocal?** Con el operador `in`: `letra in "aeiou"`.
5. **Patrón contador:** `total = 0` antes del bucle y `total = total + 1` (o `total += 1`) dentro del `if`.
6. **Casos límite:** texto vacío → 0; sin vocales → 0; los espacios no cuentan.

### Versión de referencia (comentada)

```python
def get_count(texto):
    total = 0
    for letra in texto:        # recorrer el string
        if letra in "aeiou":   # buscar con in
            total += 1         # actualizar el contador
    return total

# la misma idea, en una línea:
def get_count_corta(texto):
    return sum(1 for letra in texto if letra in "aeiou")
```

**La moraleja:** Recorrer, preguntar con `in` y actualizar un contador: es el mismo esqueleto que `has_e` y `uses_any`. Fíjate en dónde va cada pieza: `total = 0` **antes** del bucle, `total += 1` **dentro** del `if` y `return` **fuera** del bucle.
