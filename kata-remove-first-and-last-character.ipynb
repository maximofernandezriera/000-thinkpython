## El encargo

Recibes un string de **al menos 2 caracteres**. Devuelve el mismo string **sin el primer ni el último carácter**.

Ejemplos:
- `"eloquent"` → `"loquen"`
- `"country"` → `"ountr"`
- `"xyz"` → `"y"`
- `"ab"` → `""` (string vacío)

Firma: `def remove_char(texto)`.

## Hoja de ruta

1. **Entrada:** un string con 2 o más caracteres.
2. **Salida:** otro string, dos caracteres más corto.
3. **Porciones de string:** `texto[inicio:fin]` coge desde `inicio` (incluido) hasta `fin` (no incluido), como en el capítulo 8.
4. **¿Dónde empieza?** El primer carácter es el índice 0, así que hay que empezar en el 1.
5. **¿Dónde acaba?** Con índices negativos, `-1` es el último. Si `fin` es `-1`, el último se queda fuera.
6. **Strings inmutables:** no puedes borrar letras del string original. La porción crea uno nuevo.

### Versión de referencia (comentada)

```python
def remove_char(texto):
    return texto[1:-1]   # desde el índice 1 hasta antes del último

# la misma porción con índices positivos:
def remove_char2(texto):
    return texto[1:len(texto) - 1]
```

**La moraleja:** Las porciones de string funcionan igual que las de lista: `inicio` entra y `fin` no. Y como los strings son inmutables, nunca se "borra" una letra: siempre se construye un string nuevo.
