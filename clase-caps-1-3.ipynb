## 1.1 El intérprete es una calculadora que no juzga

## 1.2 Operadores aritméticos: el arsenal completo

| Operador | Significado | Ejemplo | Resultado |
|---|---|---|---|
| `+` | suma | `2 + 3` | `5` |
| `-` | resta | `7 - 10` | `-3` |
| `*` | multiplicación | `6 * 7` | `42` |
| `/` | división (¡siempre decimal!) | `84 / 2` | `42.0` |
| `//` | división entera | `84 // 5` | `16` |
| `%` | resto (módulo) | `84 % 5` | `4` |
| `**` | potencia | `2 ** 10` | `1024` |

La **precedencia** es la de matemáticas de toda la vida (PEMDAS); los paréntesis mandan.

```python
# EN VIVO: la sonda lleva 84 litros de combustible y cada maniobra gasta 5.
# ¿Cuántas maniobras completas puede hacer? ¿Cuánto combustible sobra?
84 // 5
```

```python
84 % 5
```

## 1.4 Errores: los tres sabores del fracaso

El capítulo 1 distingue tres tipos de error. Vamos a provocarlos **todos a propósito** (porque un error en clase es una lección, no un desastre):

1. **Error de sintaxis**: la frase está mal construida.
2. **Error en tiempo de ejecución (runtime)**: la frase está bien, pero explota al ejecutarse.
3. **Error semántico**: funciona... pero no hace lo que querías (¡el más peligroso!).

```python
# EN VIVO — error de sintaxis: falta cerrar el paréntesis
print("Liftoff!"
```

```python
# EN VIVO — runtime error: división entre cero
combustible = 0
84 / combustible
```

```python
# EN VIVO — error semántico: este código NO falla, pero el resultado es "incorrecto".
# Queremos la media de 10 y 20... ¿qué salió mal?
10 + 20 / 2
```

## 2.1 Variables: ponerle nombre a las cosas

Una **sentencia de asignación** crea una variable y la "apunta" a un valor. Ojo: `=` en Python no es igualdad matemática, es "ponle esta etiqueta a este valor".

```python
# EN VIVO: estado de la misión en variables
planeta_destino = "Marte"
velocidad_kmh = 40000
distancia_km = 225_000_000   # los _ son solo para leer mejor
```

```python
# EN VIVO: usar la variable devuelve su valor
velocidad_kmh
```

## 2.2 Nombres de variables: las reglas de la NASA

- Letras, dígitos y `_`, pero **sin empezar por dígito**.

- Distingue mayúsculas: `nave` ≠ `Nave`.
- No uses **palabras reservadas** (`if`, `for`, `def`, `class`...).
- Convención Python: `snake_case` (minúsculas con guiones bajos).

```python
# EN VIVO — falla a propósito: nombre ilegal
2luna = "satélite"
```

## 2.4 Expresiones vs. sentencias — y `print`

En Jupyter la última expresión de una celda se muestra sola; pero en un script `.py` no. `print()` muestra valores **siempre**, y acepta **varios argumentos** separados por comas.

```python
# EN VIVO: esto en Jupyter muestra algo; en un script .py, ¡silencio!
velocidad_kmh + 1000
```

```python
# EN VIVO: print con varios argumentos (separados por espacio automáticamente)
print("Destino:", planeta_destino, "| Velocidad:", velocidad_kmh, "km/h")
```

```python
# EN VIVO: f-strings, la forma moderna de incrustar valores en texto.
# ¡Adelanto útil aunque el libro lo explica en el cap. 3!
print(f"A {velocidad_kmh} km/h llegamos en {distancia_km // velocidad_kmh} horas")
```

# ACTO III — La tripulación trabaja (Capítulo 3) ⏰ ~55 min

## 3.1 Definir funciones: crear verbos nuevos

Una **función** empaqueta una secuencia de sentencias bajo un nombre. Definirla **no la ejecuta**: solo le enseña a Python un verbo nuevo.

```python
# EN VIVO: nuestra primera función (fíjate: al ejecutarla NO pasa nada)
def despegar():
    print("Cuenta atrás...")
    print("3... 2... 1...")
    print("¡Ignición! 🚀")
```

```python
# EN VIVO: ahora sí, la LLAMAMOS con paréntesis
despegar()
```

## 3.2 Parámetros y argumentos: funciones con entrada

Un **parámetro** es el nombre que la función usa para su entrada; el **argumento** es el valor concreto que le pasas al llamarla.

```python
# EN VIVO: una función con parámetro
def transmitir(mensaje):
    print("📡 Houston dice:", mensaje)

transmitir("Todo nominal por aquí")
```

```python
# EN VIVO: el argumento puede ser una EXPRESIÓN, se evalúa antes de entrar
transmitir("Quedan " + str(60) + " minutos de ventana")
```

## 3.3 Repetición: un bucle `for` dentro de la función

El capítulo 3 usa la repetición para mostrar la potencia de las funciones. Un `for` dentro de un `def` repite el cuerpo de la función.

```python
# EN VIVO: repetición con for — la función se ejecuta entera cada vez
def comprobar_sistemas():
    print("  ✔ Motor")
    print("  ✔ Oxígeno")
    print("  ✔ Comunicaciones")

for i in range(3):
    print(f"Comprobación nº {i + 1}")
    comprobar_sistemas()
```

## 3.4 Variables locales: lo que pasa en la función, se queda en la función

Las variables y parámetros creados **dentro** de una función son **locales**: no existen fuera. Esto nos lleva al **diagrama de pila** y al **traceback**.

```python
# EN VIVO: combustible es LOCAL a esta función
def calcular_autonomia(litros, consumo):
    horas = litros / consumo
    print(f"Autonomía: {horas} horas")

calcular_autonomia(84, 5)
```

```python
# EN VIVO — falla a propósito: `horas` no existe aquí fuera. NameError.
horas
```

**Diagrama de pila (dibujar en la pizarra):** cuando `calcular_autonomia` se ejecuta, Python crea un *frame* con sus variables locales (`litros`, `consumo`, `horas`) encima del frame principal. Al terminar, ese frame desaparece.

```python
# EN VIVO — traceback de verdad: error dentro de una función.
# Fíjate en cómo el rastro muestra LA LÍNEA de la llamada y LA LÍNEA del fallo.
def nivel_oxigeno():
    return 100 / 0      # el ingeniero escribió mal la fórmula

nivel_oxigeno()
```
