## El encargo

Recibes un texto y debes devolver `True` si tiene **el mismo número de `x` que de `o`**, y `False` si no. No distingue mayúsculas: `X` cuenta como `x` y `O` como `o`. El texto puede tener cualquier carácter.

Ejemplos:
- `"ooxx"` → **True**
- `"xooxx"` → **False**
- `"ooxXm"` → **True** (la `X` mayúscula cuenta)
- `"zpzpzpp"` → **True** (cero `x` y cero `o`: también son iguales)

Firma: `def xo(texto)`.

## Hoja de ruta

1. **Entrada:** un string con cualquier carácter.
2. **Salida:** un booleano, `True` o `False`.
3. **Dos contadores:** uno para las `x` y otro para las `o`, los dos empiezan en 0.
4. **Mayúsculas:** pasa cada letra a minúscula con `letra.lower()` antes de comparar, como hace el capítulo con `word.lower()`.
5. **La comparación final ya es un booleano:** `return equis == oes`, sin `if` ni `else`.
6. **Casos límite:** ninguna `x` ni `o` → `True`; texto vacío → `True`.

### Versión de referencia (comentada)

```python
def xo(texto):
    equis = 0
    oes = 0
    for letra in texto:
        letra = letra.lower()   # X y x cuentan igual
        if letra == "x":
            equis += 1
        elif letra == "o":
            oes += 1
    return equis == oes         # la comparación ya es True/False

# con el método count del capítulo 8:
def xo_corta(texto):
    texto = texto.lower()
    return texto.count("x") == texto.count("o")
```

**La moraleja:** Un mismo recorrido puede actualizar varias variables a la vez. Y `return a == b` devuelve directamente el booleano: no hace falta `if a == b: return True else: return False`.
