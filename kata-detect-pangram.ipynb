## El encargo

Un **pangrama** es una frase que usa todas las letras del alfabeto inglés (de la `a` a la `z`) al menos una vez. Por ejemplo, `"The quick brown fox jumps over the lazy dog"` es un pangrama.

Devuelve `True` si el texto es un pangrama y `False` si no. No importan las mayúsculas, y los números y signos de puntuación se ignoran.

Ejemplos:
- `"The quick brown fox jumps over the lazy dog"` → **True**
- `"Hola Marte"` → **False**

Firma: `def is_pangram(texto)`.

Es un kata de 6 kyu, un poco más difícil que los anteriores, pero con lo del capítulo 7 sale en pocas líneas.

### Versión de referencia (comentada)

```python
def is_pangram(texto):
    texto = texto.lower()                         # una vez, antes del bucle
    for letra in "abcdefghijklmnopqrstuvwxyz":    # recorrer lo requerido
        if letra not in texto:                    # ¿falta?
            return False                          # búsqueda: salir en cuanto falla
    return True                                   # fuera del bucle

# reutilizando la función del capítulo:
def uses_all(word, required):
    for letter in required:
        if letter not in word:
            return False
    return True

def is_pangram2(texto):
    return uses_all(texto.lower(), "abcdefghijklmnopqrstuvwxyz")
```

**La moraleja:** Muchos katas son una función del capítulo con otro nombre. Aquí el `uses_all` del libro resuelve el problema entero. El error típico es poner `return True` dentro del bucle, en un `else`: entonces la función decide mirando solo la `a`.
