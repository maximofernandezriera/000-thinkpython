En la base de Marte llegan astronautas de muchos países. Recibes el idioma de uno de ellos y debes devolver el **saludo de bienvenida** en ese idioma.


6. **Casos límite:** idioma desconocido → `"Welcome"`; el diccionario es global, no lo rellenes tú.

### Versión de referencia (comentada)

```python
SALUDOS = {
    "english": "Welcome",
    "spanish": "Bienvenido",
    "french": "Bienvenue",
    "italian": "Benvenuto",
    "german": "Willkommen",
}

def greet(idioma):
    if idioma in SALUDOS:        # ¿está la clave?
        return SALUDOS[idioma]   # leer el valor
    return "Welcome"             # por defecto

# la misma idea, en una línea, con dict.get:
def greet_get(idioma):
    return SALUDOS.get(idioma, "Welcome")
```

**La moraleja:** `d[clave]` es leer a ciegas: si la clave no está, explota. `clave in d` pregunta antes, y `d.get(clave, por_defecto)` lee con red de seguridad. Tres caras de la misma búsqueda en un diccionario.
