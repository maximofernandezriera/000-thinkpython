## El encargo

Recibes una lista de enteros. Devuelve la suma de todos **excepto el menor y el mayor** (solo una ocurrencia de cada uno).

Ejemplos:
- `[6, 2, 1, 8, 10]` → `2 + 6 + 8` = **16**
- `[1, 1, 11, 2, 3]` → `1 + 2 + 3` = **6** (se quita un 1 y el 11)

Firma: `def sum_array(lista)`. Con lista vacía o de un solo elemento devuelve `0`.

## Hoja de ruta

4. **Casos límite:** menos de 3 elementos → 0. Solo se quita UNA copia del min y del max.
5. **Ojo con `sort`:** modifica la lista in situ y devuelve None — usa `sorted(lista)` si no quieres tocarla.

### Versión de referencia (comentada)

```python
def sum_array(lista):
    if lista is None or len(lista) < 3:
        return 0
    ordenada = sorted(lista)     # sorted crea copia; sort() la tocaría
    return sum(ordenada[1:-1])   # porción sin el primero ni el último

# la otra estrategia, sin ordenar:
def sum_array2(lista):
    if lista is None or len(lista) < 3:
        return 0
    return sum(lista) - max(lista) - min(lista)
```

**La moraleja:** Dos caminos válidos: ordenar y tomar la porción `[1:-1]` (usa lo del capítulo), o restar min y max. Fíjate en `sorted()` vs `sort()`: uno copia, el otro muta y devuelve None.
