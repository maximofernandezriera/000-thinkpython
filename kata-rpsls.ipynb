## Paso 3 · Juega contra la máquina

Con tu `rpsls` ya funcionando, monta el juego. La CPU elige al azar con `random.choice`:

```
mientras elijas jugar:
    lees tu elección
    la CPU elige al azar
    rpsls(tu_eleccion, cpu) decide el ganador
    sumar puntos
```

````python
import random

def partida():
    opciones = ["rock", "paper", "scissors", "lizard", "spock"]
    emoji = {"rock": "🪨", "paper": "📄", "scissors": "✂️", "lizard": "🦎", "spock": "🖖"}
    yo = cpu = 0
    while True:
        tuya = input("Elige (rock/paper/scissors/lizard/spock) o 'salir': ").lower().strip()
        if tuya == "salir":
            break
        if tuya not in opciones:
            print("Elección no válida."); continue
        suya = random.choice(opciones)
        resultado = rpsls(tuya, suya)
        print(f"  Tú: {emoji[tuya]}  ·  CPU: {emoji[suya]}  →  {resultado}")
        if resultado == "Player 1 Won!": yo += 1
        elif resultado == "Player 2 Won!": cpu += 1
        print(f"  Marcador: tú {yo} — CPU {cpu}")
    print(f"\nFinal: tú {yo} — CPU {cpu}. ¡Bazinga!")

# Descomenta para jugar:
# partida()
````

La mayoría de soluciones de Codewars encadenan 10 `if`. El diccionario de conjuntos:
