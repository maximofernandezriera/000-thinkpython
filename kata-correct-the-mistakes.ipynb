## El encargo

El programa que digitaliza textos antiguos se equivoca con tres caracteres:

- la `S` se lee como `5`;
- la `O` se lee como `0`;
- la `I` se lee como `1`.

Recibes el texto digitalizado y debes devolverlo **corregido**. Los números del texto siempre son errores.

Ejemplos:
- `"L0ND0N"` → `"LONDON"`
- `"DUBL1N"` → `"DUBLIN"`
- `"51NGAP0RE"` → `"SINGAPORE"`

Firma: `def correct(texto)`.

## Hoja de ruta

1. **Entrada:** un string en mayúsculas que puede tener `5`, `0` y `1`.
2. **Salida:** el string corregido.
3. **Método `replace`:** `texto.replace(viejo, nuevo)` cambia todas las apariciones, como en el apartado «Find and replace» del capítulo 8.
4. **Ojo, inmutables:** `replace` **no** cambia el string original, devuelve uno nuevo. Hay que guardarlo: `texto = texto.replace(...)`.
5. **Tres errores, tres reemplazos:** se pueden encadenar uno detrás de otro.
6. **Casos límite:** texto sin errores → igual; texto vacío → vacío.

### Versión de referencia (comentada)

```python
def correct(texto):
    texto = texto.replace("5", "S")   # replace devuelve un string NUEVO
    texto = texto.replace("0", "O")   # por eso hay que guardarlo
    texto = texto.replace("1", "I")
    return texto

# encadenado: cada replace trabaja sobre el resultado del anterior
def correct_corta(texto):
    return texto.replace("5", "S").replace("0", "O").replace("1", "I")
```

**La moraleja:** Los métodos de string no cambian el string: devuelven uno nuevo. Llamar a `texto.replace(...)` sin guardar el resultado es un error muy habitual, y no da ningún mensaje de error: simplemente no cambia nada.
