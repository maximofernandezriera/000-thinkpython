### Una lista es una secuencia

Como un string, una **lista** es una secuencia de valores. La diferencia: en un string los valores son caracteres; en una lista pueden ser de cualquier tipo. Cada valor es un **elemento**, y la lista se escribe entre corchetes.

```python
tripulacion = ["Ada", "Neil", "Valentina", "Yuri"]
tripulacion
```

```python
# las temperaturas (en °C) de los primeros sols en el cráter Jezero
temperaturas = [-63, -58, -71, -49, -66]
len(temperaturas)
```

Los elementos no tienen por qué ser del mismo tipo, e incluso puede haber una lista dentro de otra (una lista **anidada**). Antes de ejecutar: ¿cuántos elementos tiene `muestra`?

```python
# nombre, masa en gramos, ¿tiene agua?, coordenadas [x, y] del hallazgo
muestra = ["basalto", 12.5, True, [18, 77]]
len(muestra)
```

Cuatro: la lista de coordenadas cuenta como **un solo** elemento. Y la lista vacía existe, con longitud 0. Es el punto de partida de casi todos los bucles que construyen listas.

```python
hallazgos = []
len(hallazgos)
```

### Las listas son mutables

Se accede a un elemento con corchetes y el índice **empieza en 0**, igual que en los strings. Los índices negativos cuentan desde el final.

```python
tripulacion[0], tripulacion[-1]
```

La gran diferencia con los strings: las listas se pueden **modificar**. Si el corchete aparece a la izquierda de una asignación, cambia ese elemento. Neil se queda en órbita y baja Mae en su lugar:

```python
tripulacion[1] = "Mae"
tripulacion
```

### Métodos de lista

Las listas traen métodos propios. Los cuatro que más vais a usar:

- `append(x)`: añade un elemento al final.
- `extend(otra)`: añade todos los elementos de otra lista.
- `pop(i)`: quita el elemento de la posición `i` y lo **devuelve**.
- `remove(x)`: quita la primera aparición del valor `x`.

```python
hallazgos = []
hallazgos.append("basalto")
hallazgos.append("olivino")
hallazgos
```

```python
hallazgos.extend(["hematita", "yeso", "arcilla"])
hallazgos
```

```python
# el rover analiza la primera muestra: pop la quita y nos la entrega
analizada = hallazgos.pop(0)
analizada, hallazgos
```

```python
hallazgos.remove("yeso")
hallazgos
```

```python
# ¿y si quitamos algo que no está?
hallazgos.remove("oro")
```

`ValueError: list.remove(x): x not in list`. Si no estáis seguros, comprobad antes con `in`.

Fijaos en un detalle que vuelve en la sección de depuración: `append`, `extend` y `remove` **modifican** la lista y devuelven `None`. Los métodos de string hacen lo contrario: dejan el original intacto y devuelven uno nuevo.

### Recorrer una lista con un bucle

El `for` del capítulo 7 recorre una lista exactamente igual que recorría un string: un elemento por vuelta.

```python
for persona in tripulacion:
    print("Traje presurizado:", persona)
```

Y el patrón *inicializar, recorrer, actualizar* sigue funcionando. Sin usar `min`: ¿cuál ha sido el sol más frío? Guardamos el mínimo visto hasta ahora y lo actualizamos cuando aparece uno menor.

```python
def mas_frio(temps):
    minimo = temps[0]          # inicializar con el primero
    for t in temps:
        if t < minimo:         # actualizar si encontramos uno menor
            minimo = t
    return minimo

mas_frio(temperaturas)
```

Seguid el bucle paso a paso: fijaos en cómo `t` toma cada temperatura y `minimo` solo cambia cuando aparece una más baja (-63 → -71).

El mismo patrón sirve para **construir** una lista nueva: empezar con `[]` y hacer `append` dentro del bucle. Por ejemplo, convertir las temperaturas a kelvin:

```python
def a_kelvin(temps):
    resultado = []
    for t in temps:
        resultado.append(round(t + 273.15, 2))   # round: dos decimales
    return resultado

a_kelvin(temperaturas)
```

```python
# un for sobre una lista vacía no ejecuta nunca el cuerpo
for x in []:
    print("Esto no se imprime nunca")
```

### Objetos y valores

Ahora viene la parte del capítulo que más errores evita en la vida real. ¿Qué creéis que da cada una de estas dos celdas?

```python
a = [1, 2, 3]
b = [1, 2, 3]
a == b
```

```python
a is b
```

`==` compara **valores**: las dos listas tienen los mismos elementos, así que son **equivalentes**. `is` pregunta si son **el mismo objeto**, y no lo son: hemos creado dos listas distintas que casualmente contienen lo mismo. Equivalentes, pero no **idénticas**.

### Aliasing

Ahora en lugar de crear dos listas asignamos una variable a otra. Antes de ejecutar: después de la tercera línea, ¿qué vale `equipo_a`?

```python
equipo_a = ["Ada", "Mae"]
equipo_b = equipo_a          # NO copia la lista: crea otro nombre para la misma
equipo_b.append("Yuri")
equipo_a
```

```python
equipo_a is equipo_b
```

`equipo_b = equipo_a` no crea una lista nueva: crea **otro nombre** para la misma lista. Cada asociación entre una variable y un objeto es una **referencia**; cuando un objeto tiene dos nombres decimos que tiene un **alias**. Como la lista es mutable, lo que se cambia por un nombre se ve por el otro.

El esquema de la pizarra:

Si lo que queremos es una copia independiente, se pide explícitamente con la porción completa `[:]` o con `list()`:

```python
equipo_a = ["Ada", "Mae"]
equipo_c = equipo_a[:]       # copia de verdad
equipo_c.append("Valentina")
equipo_a, equipo_c
```

Con objetos inmutables, como los strings, el aliasing no da problemas: nadie puede cambiar el objeto compartido.

### Listas como argumentos

Cuando pasamos una lista a una función, la función recibe una **referencia** a la misma lista. Es otro alias: el parámetro y la variable original son dos nombres para el mismo objeto. Si la función modifica la lista, el cambio se queda después de que la función termine.

```python
def analizar_siguiente(cola):
    return cola.pop(0)

cola_muestras = ["basalto", "olivino", "hematita"]
analizar_siguiente(cola_muestras)
```

```python
cola_muestras     # ¿qué queda en la cola?
```

### La lista de palabras del libro

En el capítulo 7 leíamos `words.txt` entero cada vez que buscábamos algo. Es mucho más eficiente leerlo **una vez** y guardar las palabras en una lista.

```python
from os.path import basename, exists

def descargar(url):
    archivo = basename(url)
    if not exists(archivo):
        from urllib.request import urlretrieve
        urlretrieve(url, archivo)
    return archivo

descargar('https://raw.githubusercontent.com/AllenDowney/ThinkPython/v3/words.txt')
```

```python
# versión 1: inicializar con [] y append en cada vuelta
palabras = []
for linea in open('words.txt'):
    palabras.append(linea.strip())

len(palabras)
```

```python
# versión 2: leer todo el archivo como un string y partirlo con split
palabras = open('words.txt').read().split()
len(palabras)
```

```python
"mars" in palabras, "marte" in palabras    # es una lista de palabras en inglés
```

### Depuración: la mayoría de métodos de lista devuelven None

Con strings estamos acostumbrados a escribir `s = s.strip()`, porque el método devuelve un string nuevo. Con listas, la tentación es escribir lo mismo:

```python
cola = ["basalto", "olivino", "hematita"]
cola = cola.remove("olivino")    # ¡MAL!
print(cola)
```

```python
# la siguiente operación con la lista revienta
cola.append("yeso")
```

`AttributeError: 'NoneType' object has no attribute 'append'`. `remove` modificó la lista y devolvió `None`, y al asignarlo hemos perdido la lista. Cuando veáis un `NoneType` donde esperabais una lista, buscad un método de lista cuyo resultado se haya asignado. Lo correcto es llamar al método sin asignar:

```python
cola = ["basalto", "olivino", "hematita"]
cola.remove("olivino")
cola
```

Comparad las dos versiones en PythonTutor: en la mala, después de la asignación `cola` deja de apuntar a la lista y pasa a valer `None`; en la buena, la lista sigue ahí, con un elemento menos.
