- `[1, 1, 2]` → **2**

## Hoja de ruta

1. **Entrada:** una lista de enteros.
2. **Salida:** el entero que aparece un número impar de veces.
3. **Fase 1 — contar:** el mismo diccionario `{numero: cuenta}` del kata anterior (`cuentas[n] = cuentas.get(n, 0) + 1`).
4. **Fase 2 — buscar:** recorre el diccionario (`for n in cuentas:`) y devuelve la clave cuya cuenta sea impar: `cuentas[n] % 2 == 1`.
5. **El impar con `%**:` un número es impar si `n % 2 == 1`, lo viste en el capítulo 5.
6. **Casos límite:** un solo elemento → ese elemento; el impar puede ser cualquier número, también 0.

### Versión de referencia (comentada)

```python
def find_it(secuencia):
    cuentas = {}
    for n in secuencia:                    # fase 1: contar
        cuentas[n] = cuentas.get(n, 0) + 1
    for n in cuentas:                      # fase 2: buscar el impar
        if cuentas[n] % 2 == 1:
            return n

# truco avanzado con XOR: los pares se anulan a sí mismos
def find_it_xor(secuencia):
    resultado = 0
    for n in secuencia:
        resultado = resultado ^ n
    return resultado
```

**La moraleja:** Dos fases con nombre propio: **contar** (diccionario frecuencia) y **filtrar** (recorrer el diccionario preguntando por cada cuenta). El `for n in cuentas` recorre las claves; `cuentas[n]` te da el valor.
