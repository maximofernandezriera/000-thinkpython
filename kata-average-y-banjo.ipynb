Ejemplos:

## Hoja de ruta

5. **Casos límite:** `[]` → 0; un solo elemento → ese mismo número; con negativos se suma igual.

```python
def find_average(numeros):
    if len(numeros) == 0:      # guardia: sin esto, dividir explota
        return 0
    total = 0
    for n in numeros:          # acumulador de siempre
        total += n
    return total / len(numeros)

# con sum(), lo mismo en dos líneas:
def find_average_corta(numeros):
    if len(numeros) == 0:
        return 0
    return sum(numeros) / len(numeros)
```

**La moraleja:** El acumulador otra vez: `total` **antes** del bucle, `+=` **dentro**, `return` **fuera**. Y una regla de oro: antes de dividir, asegúrate de que el divisor no es cero. `sum()` ya viene de fábrica, pero la guardia sigue siendo tuya.

### Versión de referencia (comentada)

```python
def are_you_playing_banjo(nombre):
    if nombre[0] in "Rr":                    # cubre R y r a la vez
        return nombre + " plays banjo"
    return nombre + " does not play banjo"   # sin else: si llega aquí, no tocaba banjo

# igual, normalizando a minúscula:
def are_you_playing_banjo_lower(nombre):
    if nombre[0].lower() == "r":
        return nombre + " plays banjo"
    return nombre + " does not play banjo"

# con f-string, eligiendo primero la coletilla:
def are_you_playing_banjo_f(nombre):
    coletilla = " plays banjo" if nombre[0] in "Rr" else " does not play banjo"
    return f"{nombre}{coletilla}"
```

**La moraleja:** `nombre[0]` es la lupa sobre el primer carácter y `in "Rr"` te ahorra escribir dos comparaciones. Fíjate en el `return` sin `else`: cuando el `if` devuelve, lo que queda debajo es el camino "no". Y cuidado con el espacio: `+ " plays"` no es lo mismo que `+ "plays"`.
